```python
new_description=[
    "smart tv with hdr support"
   ]
new_description_vectorized=vectorizer.transform(new_description)
predictions=model.predict(new_description_vectorized)
for description,pred in zip(new_description,predictions):
    print(f"Description: {description}\nPredicted Category: {pred}\n")
```
